# 🎓 Student Performance Prediction

## 📌 Objective

The objective of this project is to analyze student performance data and build a Machine Learning model that predicts students' mathematics scores based on demographic, educational, and social factors.

```python
# Import Libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_style("whitegrid")

import warnings
warnings.filterwarnings('ignore')
```

```python
# Load DataSet
df = pd.read_csv("../data/raw/StudentsPerformance.csv")
df.head()
```

```python
df.drop_duplicates(inplace=True)
```

```python
# Check Data Types
df.dtypes
```

```python
# Unique Values
for col in df.columns:
    print(col)
    print(df[col].unique())
    print("-"*50)
```

```python
# Correlation Analysis
numeric_df = df.select_dtypes(include=np.number)

plt.Figure(figsize=(8,6))

sns.heatmap(
    numeric_df.corr(),
    annot= True,
    cmap= "coolwarm"
)

plt.show()
```

```python
# Distribution Plots
numeric_df.hist(
    figsize=(12,8),
    bins=20
)

plt.show()
```

```python
# Boxplots (Outlier Analysis)
plt.figure(figsize=(8,5))

sns.boxplot(x=df['math score'])

plt.title('Math Score Distribution')

plt.show()
```

```python
# Boxplots (Outlier Analysis)
plt.figure(figsize=(8,5))

sns.boxplot(x=df['reading score'])

plt.title('Reading Score Distribution')

plt.show()
```

```python
# Boxplots (Outlier Analysis)
plt.figure(figsize=(8,5))

sns.boxplot(x=df['writing score'])

plt.title('Writing Score Distribution')

plt.show()
```

```python
# Pair Plot
sns.pairplot(numeric_df)
plt.show()
```

# Findings
- Dataset contains 1000 records.
- No missing values.
- No Duplicat records.
- Math, reading, and writing scores are strongly correlated.
- Numerical features have limited outliers.
- Categorical feature require encoding before modeling
