# Loan Approval Prediction

## Data Preprocessing

In this notebook, we prepare the raw dataset for machine learning.

The preprocessing pipeline includes:

- Loading the raw dataset
- Feature selection
- Removing unnecessary columns
- Encoding categorical variables
- Preparing features and target
- Train-test splitting
- Saving the processed dataset

The output of this notebook will be used in the model training notebook.

```python
# Import Libraries
import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder
```

```python
# Load DataSet
df = pd.read_csv("../data/raw/loan_approval_dataset.csv")

df.head()
```

```python
# DataSet Shape
print(f"Rows    : {df.shape[0]}")
print(f"Columns : {df.shape[1]}")
```

```python
# Column Names
df.columns
```

## Why are we removing extra spaces from column names?

Some datasets contain accidental leading or trailing spaces in their column names.

These spaces can cause errors during feature selection and model deployment.

Using `str.strip()` standardizes all column names and makes the dataset easier to work with.

```python
df.columns = df.columns.str.strip()

df.columns
```

```python
# Checking missing values
df.isnull().sum()
```

```python
# Check Duplicate values
df.duplicated().sum()
```

## Why are we removing Loan ID?

`loan_id` is a unique identifier assigned to each loan application.

It does not contain any meaningful information for predicting loan approval and may introduce unnecessary noise into the model.

Removing identifier columns is a common preprocessing practice in machine learning.

```python
df.drop("loan_id", axis=1, inplace=True)

df.head()
```

```python
df.shape
```

```python
categorical_columns = df.select_dtypes(include="object").columns

categorical_columns
```

## Why do we encode categorical features?

Machine learning algorithms work with numerical data.

Categorical features such as "Graduate" and "Not Graduate" must be converted into numerical values before model training.

Since our categorical features contain only two categories each, Label Encoding is an appropriate and simple choice.

```python
# Label Encode Features
encoder = LabelEncoder()

df["education"] = encoder.fit_transform(df["education"])

df["self_employed"] = encoder.fit_transform(df["self_employed"])

df.head()
```

## Why do we encode the target variable?

The target column contains text values ("Approved" and "Rejected").

Classification algorithms require numerical labels.

We convert:

- Approved → 0
- Rejected → 1

This allows the model to learn and predict loan approval outcomes.

```python
target_encoder = LabelEncoder()

df["loan_status"] = target_encoder.fit_transform(df["loan_status"])

df.head()
```

```python
# Verify Encoding
df.head()
```

```python
df.dtypes
```

```python
# Features and Target
X = df.drop("loan_status", axis=1)

y = df["loan_status"]

print("Features Shape :", X.shape)
print("Target Shape   :", y.shape)
```

## Why do we split the dataset?

To evaluate the model fairly, we divide the data into:

- Training Set (80%)
- Testing Set (20%)

The model learns from the training data and is evaluated on unseen testing data.

Using `stratify=y` ensures that both sets maintain the same proportion of approved and rejected loan applications.

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)
```

```python
# Check Shapes
print("Training Features :", X_train.shape)
print("Testing Features  :", X_test.shape)

print("Training Target   :", y_train.shape)
print("Testing Target    :", y_test.shape)
```

## Why do we create a processed dataset?

After completing preprocessing, we save the cleaned dataset for future use.

This ensures that the model training notebook always uses the same processed data, making the workflow organized, reproducible, and efficient.

```python
processed_df = pd.concat([X, y], axis=1)

processed_df.head()
```

```python
# Save Processed DataSet
processed_df.to_csv(
    "../data/processed/processed_loan.csv",
    index=False
)

print("✅ Processed dataset saved successfully.")
```

```python
# Verify Saved Dataset
pd.read_csv("../data/processed/processed_loan.csv").head()
```

# Preprocessing Summary

The following preprocessing steps were completed:

- Cleaned column names by removing extra spaces.
- Verified that the dataset contains no missing values.
- Verified that there are no duplicate records.
- Removed the `loan_id` column as it is not useful for prediction.
- Encoded categorical features (`education` and `self_employed`).
- Encoded the target variable (`loan_status`).
- Split the dataset into training and testing sets using stratified sampling.
- Saved the processed dataset for model training.
