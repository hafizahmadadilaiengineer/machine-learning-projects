```python
# import libraries
import pandas as pd
import numpy as np
import joblib

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
```

```python
# Load Dataset
df = pd.read_csv("../data/raw/heart.csv")
df.head()
```

```python
print('Shape Before: ', df.shape)

df = df.drop_duplicates()

print("Shape After: ", df.shape)
```

```python
# Split Feature and Target
X = df.drop('target', axis= 1)

y = df['target']
```

```python
# Train Test Split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = 0.20, random_state = 42, stratify = y)
```

```python
# Feature Scaling
scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train)

X_test_scaled = scaler.transform(X_test)
```

```python
# Save the Scaler
joblib.dump(scaler, "../models/scler.pkl")

print("Scaler Saved Sucessfully.")
```

```python
# Save Processed Dataset
processed_data = pd.concat([X,y], axis = 1)

processed_data.to_csv("../data/processed/processed_heart.csv", index=False)

print("Processed Dataset Saved Sucessfully.")
```

# Observations

- Duplicate rows were removed to improve data quality and prevent model bias.
- Features and target were separated for supervised learning.
- The dataset was split into training and testing sets using an 80:20 ratio.
- Stratified sampling was used to preserve the original class distribution in both datasets.
- Numerical features were standardized using StandardScaler for distance-based algorithms such as K-Nearest Neighbors.
- The scaler was fitted only on the training data and applied to the test data to prevent data leakage.
- The processed dataset and scaler were saved for use during model training and deployment.
