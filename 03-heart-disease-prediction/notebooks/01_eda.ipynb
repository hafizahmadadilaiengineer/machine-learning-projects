```python
# Import Libraries
import pandas as pd
import numpy as np

import matplotlib.pyplot as plt
import seaborn as sns

# Display settings
pd.set_option("Display.max_columns", None)

# Plot Style
sns.set_style("whitegrid")

print("Libraries imported sucessfully.")
```

```python
# Load Dataset
df = pd.read_csv("../data/raw/heart.csv")
df.head()
```

```python
# Dataset Shape
print("Rows :", df.shape[0])
print("Columns :", df.shape[1])
```

```python
# Dataset Information
df.info()
```

```python
# Statisticl Summary 
df.describe().T
```

```python
# Missing Values
df.isnull().sum()
```

```python
# Duplicated Values
df.duplicated().sum()
```

```python
# Target Distribution
plt.Figure(figsize=(6,4))

sns.countplot(data=df, x="target", palette="Set2")

plt.title("Target Variable Distribution")
plt.xlabel("Heart Disease")
plt.ylabel("Count")

plt.show()
```

```python
# Correlation Heatmap
plt.Figure(figsize=(12,8))

sns.heatmap(df.corr(), cmap="coolwarm", annot=False)

plt.title("Correlation Heatmap")

plt.show()
```

```python
# Numerical Distribution
df.hist(figsize=(18,15))

plt.tight_layout()

plt.show()
```

```python
# Boxplots
plt.figure(figsize=(18,12))

for i, column in enumerate(df.columns[:-1],1):

    plt.subplot(4,4,i)

    sns.boxplot(y=df[column])

    plt.title(column)

plt.tight_layout()

plt.show()
```

# Observations

- The dataset contains **1,025 rows** and **14 columns**, including 13 input features and 1 target variable.
- No missing values were found, so data imputation is not required.
- A large number of duplicate rows (**723**) were detected. These duplicates will be removed during preprocessing to improve data quality and prevent model bias.
- The target variable is approximately balanced, making it suitable for training classification models without additional balancing techniques.
- Outliers were observed in features such as **chol**, **trestbps**, **ca**, and **oldpeak**. Since these represent valid medical measurements rather than data entry errors, they will be retained.
