# Customer Churn Prediction System

## Data Preprocessing

In this notebook, we will clean and prepare the dataset for machine learning by handling missing values, removing unnecessary features, encoding categorical variables, splitting the data, and saving the processed dataset.

## Step 1: Import Libraries

Import the required libraries for data preprocessing.

```python
# Data Manipulation
import pandas as pd
import numpy as np

# Data Preprocessing
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder

print("Libraries imported successfully.")
```

## Step 2: Load Dataset

Load the raw dataset for preprocessing.

```python
df = pd.read_csv("../data/raw/telco.csv")

df.head()
```

## Step 3: Display Dataset Information

Check the dataset structure before preprocessing.

```python
df.info()
```

## Step 4: Check Missing Values

Identify the columns containing missing values.

```python
missing_values = df.isnull().sum()

missing_values[missing_values > 0]
```

## Step 5: Handle Missing Values

Fill missing values in business-related features and remove columns that introduce data leakage.

```python
# Fill Missing Values
df["Offer"] = df["Offer"].fillna("No Offer")

df["Internet Type"] = df["Internet Type"].fillna("No Internet")

# Drop Leakage Columns
df.drop(
    columns=[
        "Churn Category",
        "Churn Reason"
    ],
    inplace=True
)

print("Missing values handled successfully.")
```

## Step 6: Verify Missing Values

Verify that all missing values have been handled successfully.

```python
df.isnull().sum().sum()
```

## Step 7: Remove Unnecessary Features

Remove identifier columns and features that can cause data leakage during model training.

```python
# Drop Unnecessary Columns
df.drop(
    columns=[
        "Customer ID",
        "Customer Status",
        "Churn Score",
        "CLTV"
    ],
    inplace=True
)

print("Unnecessary columns removed successfully.")
```

## Step 8: Encode the Target Variable

Convert the target variable into numerical values for machine learning.

```python
# Encode Target Variable
label_encoder = LabelEncoder()

df["Churn Label"] = label_encoder.fit_transform(df["Churn Label"])

df["Churn Label"].head()
```

## Step 9: Encode Categorical Features

Convert categorical features into numerical values using Label Encoding.

```python
# Encode Categorical Features
categorical_columns = df.select_dtypes(include="object").columns

for column in categorical_columns:
    df[column] = LabelEncoder().fit_transform(df[column])

print("Categorical features encoded successfully.")
```

```python
df.info()
```

## Step 10: Split Features and Target

Separate the independent features (X) and the target variable (y).

```python
# Features and Target
X = df.drop(columns="Churn Label")

y = df["Churn Label"]

print("Features Shape:", X.shape)
print("Target Shape:", y.shape)
```

## Step 11: Split the Dataset

Split the dataset into training and testing sets.

We use an 80:20 split and apply stratification to preserve the class distribution in both training and testing datasets.

```python
# Train-Test Split
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training Features:", X_train.shape)
print("Testing Features:", X_test.shape)
print("Training Target:", y_train.shape)
print("Testing Target:", y_test.shape)
```

## Step 12: Save the Processed Dataset

Save the cleaned dataset for future use in model training and deployment.

```python
# Save Processed Dataset
df.to_csv(
    "../data/processed/processed_telco.csv",
    index=False
)

print("Processed dataset saved successfully.")
```

## Key Findings

- Missing values were successfully handled.
- Business-related missing values were filled with meaningful categories.
- Data leakage features were removed.
- Customer ID was removed because it is only an identifier.
- The target variable was label encoded.
- All categorical features were converted into numerical values.
- The dataset was split into features and target.
- The processed dataset was saved for model training.
